```python
from pathlib import Path

DATA_DIR = Path("data/yolo/russian_plate")
for split in ("train", "val", "test"):
    images = list((DATA_DIR / "images" / split).glob("*"))
    labels = list((DATA_DIR / "labels" / split).glob("*.txt"))
    print(f"{split:5s}: {len(images):6d} images, {len(labels):6d} labels")

print((DATA_DIR / "data.yaml").read_text())
```

```python
import random
from PIL import Image, ImageDraw
import matplotlib.pyplot as plt

split = "train"
image_path = random.choice(list((DATA_DIR / "images" / split).glob("*")))
label_path = DATA_DIR / "labels" / split / f"{image_path.stem}.txt"

image = Image.open(image_path).convert("RGB")
draw = ImageDraw.Draw(image)
width, height = image.size

for line in label_path.read_text().splitlines():
    cls, x, y, w, h = map(float, line.split()[:5])
    left = (x - w / 2) * width
    top = (y - h / 2) * height
    right = (x + w / 2) * width
    bottom = (y + h / 2) * height
    draw.rectangle((left, top, right, bottom), outline="red", width=3)

plt.figure(figsize=(10, 6))
plt.imshow(image)
plt.axis("off")
plt.title(str(image_path.name));
```
